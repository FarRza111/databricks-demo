```python
import pandas as pd
from dotenv import load_dotenv

load_dotenv()
```

Getting Connection String to Databricks Warehouse


```python
from databricks import sql
import os

with sql.connect(server_hostname = os.getenv("DATABRICKS_SERVER_HOSTNAME"),
                 http_path       = os.getenv("DATABRICKS_HTTP_PATH"),
                 access_token    = os.getenv("DATABRICKS_TOKEN")) as connection:
    
    with connection.cursor() as cursor:
        cursor.execute("SELECT * FROM analytics.jaffle_shop.fct_customer_orders LIMIT 10")
        result = cursor.fetchall()

        columns = [desc[0] for desc in cursor.description]  
        
        df = pd.DataFrame(result, columns=columns)
        print(df)
        
        # for row in result:
        #     print(row)
```

Create Post Statement for Query 

```python
import requests


DATABRICKS_INSTANCE = ""
DATABRICKS_TOKEN = ""
WAREHOUSE_ID = ""


url = f"https://{DATABRICKS_INSTANCE}/api/2.0/sql/statements"

headers = {
    "Authorization": f"Bearer {DATABRICKS_TOKEN}",
    "Content-Type": "application/json"
}

payload = {
    "statement": "SELECT * FROM analytics.jaffle_shop.fct_customer_orders LIMIT 20",
    "wait_timeout": "10s",
    "warehouse_id": WAREHOUSE_ID,
    "disposition": "INLINE"
}

response = requests.post(url, headers=headers, json=payload)

print("Status code:", response.status_code)
print("Response body:", response.json())
```
